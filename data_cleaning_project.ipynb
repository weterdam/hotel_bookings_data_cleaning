```python
# import the required libraries
import pandas as pd
import numpy as np
```

```python
# read in the file 
hotel_bookings = pd.read_csv('hotel_bookings.csv')
hotel_bookings.head()
```

```python
# Find the percentage of missing values in each column
hotel_bookings.isnull().sum()*100/len(hotel_bookings)
```

```python
# Deal with the missing values 
hotel_bookings = hotel_bookings.drop('company',axis=1)

hotel_bookings.loc[hotel_bookings['children'].isnull()]['children']
hotel_bookings['children'] = hotel_bookings['children'].replace(np.nan,-1)
hotel_bookings['country'] = hotel_bookings['country'].replace(np.nan,'Unknown')
mean = hotel_bookings['agent'].mean()
hotel_bookings['agent'] = hotel_bookings['agent'].replace(np.nan,mean)
```

```python
# Rename columns 
hotel_bookings.rename(columns = {'adults' : 'num_adults','children' : 'num_children','babies' : 'num_babies'},inplace=True)
hotel_bookings.columns
```

```python
# Correct data types
hotel_bookings.dtypes
hotel_bookings = hotel_bookings.astype({'num_children' : 'int','is_repeated_guest': 'boolean','is_canceled' : 'boolean'})
hotel_bookings.dtypes
```

```python
# Create bins
hotel_bookings['lead_time'].unique()
bins = [0,100,200,300,400,500,600,700,800]
labels = ['0-100','101-200','201-300','301-400','401-500','501-600','601-700','701-800']
hotel_bookings['lead_time_binned'] = pd.cut(hotel_bookings['lead_time'],bins=bins,labels=labels)
hotel_bookings[['lead_time','lead_time_binned']]
```

```python
# Handle duplicates in the data
hotel_bookings.loc[hotel_bookings.duplicated(keep=False)]
hotel_bookings = hotel_bookings.drop_duplicates(keep='last')
hotel_bookings.shape
```
